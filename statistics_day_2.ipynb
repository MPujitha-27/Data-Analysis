Why Missing Values Are Replaced by Central Tendencies
1. To Avoid Losing Data

If you drop rows with missing values, you may lose a large portion of data.
Replacing them with mean/median/mode helps keep all rows for analysis and model training.

2. To Maintain Statistical Balance

Central tendency values (mean, median, mode) represent the typical or most common value in a dataset.
So filling missing values with them helps preserve the overall distribution of the data.

3. To Improve Model Performance

Most machine-learning models cannot handle missing values directly.
Imputing them allows the model to use the full dataset and improves accuracy.

4. To Keep the Dataset Simple and Consistent

Central tendency methods are:

Easy to calculate

Fast

Work well when missing values are few

Do not require complex assumptions

This makes them a popular choice, especially for beginners and small datasets.

5. Because Mean/Median/Mode Are Good Estimates

Mean → for continuous data with normal distribution

Median → for skewed or outlier-heavy data

Mode → for categorical values

These estimates often represent the value you would expect to see if there was no missing value.

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\kopur\Downloads\loan_approved.csv")

for col in df.columns:
    if df[col].dtype == 'object':        # Categorical
        df[col] = df[col].fillna(df[col].mode()[0])
    else:                                # Numeric
        df[col] = df[col].fillna(df[col].median())

print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(r"C:\Users\kopur\Downloads\loan_approved.csv")

# Calculate Z-score for ApplicantIncome
df["ApplicantIncome_zscore"] = (df["ApplicantIncome"] - df["ApplicantIncome"].mean()) / df["ApplicantIncome"].std()

# Define threshold (commonly 3 or -3)
threshold = 3

# Detect outliers
outliers = df[np.abs(df["ApplicantIncome_zscore"]) > threshold]

print("Outliers in ApplicantIncome (using Z-score):")
print(outliers[["ApplicantIncome", "ApplicantIncome_zscore"]])
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(r"C:\Users\kopur\Downloads\loan_approved.csv")

# Function to detect outliers using Z-score
def detect_outliers_zscore(column):
    z = (df[column] - df[column].mean()) / df[column].std()
    outliers = df[np.abs(z) > 3]
    return outliers[[column]]

# 1. ApplicantIncome Outliers
print("\n===== ApplicantIncome Outliers =====")
applicant_outliers = detect_outliers_zscore("ApplicantIncome")
print(applicant_outliers)

# 2. CoapplicantIncome Outliers
print("\n===== CoapplicantIncome Outliers =====")
coapplicant_outliers = detect_outliers_zscore("CoapplicantIncome")
print(coapplicant_outliers)

# 3. LoanAmount Outliers
print("\n===== LoanAmount Outliers =====")
loanamount_outliers = detect_outliers_zscore("LoanAmount")
print(loanamount_outliers)
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(r"C:\Users\kopur\Downloads\loan_approved.csv")

# Columns you want to analyze
cols = ["ApplicantIncome", "CoapplicantIncome", "LoanAmount", "Loan_Amount_Term"]

# --- Functions ---

# Skewness classification
def skew_cutoff(value):
    if abs(value) < 0.5:
        return "Normal (Acceptable)"
    elif abs(value) < 1:
        return "Moderately Skewed (Acceptable)"
    else:
        return "Highly Skewed (Not Normal)"

# Kurtosis classification
# Pearson kurtosis: Normal distribution ≈ 3
def kurtosis_cutoff(value):
    if 2 <= value <= 4:
        return "Normal (Acceptable)"
    elif value < 2:
        return "Platykurtic (Light tails)"
    else:
        return "Leptokurtic (Heavy tails - Not Normal)"

# Classification summary for normality
def classify(skew_cat, kurt_cat):
    if "Not Normal" in skew_cat or "Leptokurtic" in kurt_cat:
        return "NOT NORMAL"
    return "NORMAL"

# --- Perform calculations and print results ---

print("\n===== SKEWNESS & KURTOSIS REPORT =====\n")

for col in cols:
    skew_val = df[col].skew()
    kurt_val = df[col].kurtosis() + 3   # Convert Fisher to Pearson kurtosis
    
    skew_result = skew_cutoff(skew_val)
    kurt_result = kurtosis_cutoff(kurt_val)
    final_status = classify(skew_result, kurt_result)

    print(f"Column: {col}")
    print(f"  Skewness = {skew_val:.4f}  →  {skew_result}")
    print(f"  Kurtosis = {kurt_val:.4f}  →  {kurt_result}")
    print(f"  Final Classification: {final_status}")
    print("-" * 60)
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.preprocessing import PowerTransformer
import matplotlib.pyplot as plt

# Columns to transform (LoanAmount removed)
cols = ["ApplicantIncome", "CoapplicantIncome", "Loan_Amount_Term"]

pt = PowerTransformer(method="yeo-johnson")

for col in cols:
    print("\n==============================")
    print(f"Processing Column: {col}")
    print("==============================")

    # Yeo-Johnson transformation
    transformed = pt.fit_transform(data[[col]])

    # Show shape
    print("Shape:", transformed.shape)

    # Skewness & Kurtosis after Yeo-Johnson
    skew_val = pd.Series(transformed.reshape(len(data),)).skew()
    kurt_val = pd.Series(transformed.reshape(len(data),)).kurtosis()

    print("Yeo-Johnson Skewness:", skew_val)
    print("Yeo-Johnson Kurtosis:", kurt_val)

    # Plot histogram
    sns.histplot(transformed, kde=True)
    plt.title(f"Histogram after PowerTransform: {col}")
    plt.show()

    # Log transform — only if values > 0
    if (data[col] > 0).all():
        log_trans = np.log(data[col])
        print("Log Skewness:", log_trans.skew())
        print("Log Kurtosis:", log_trans.kurtosis())

        sns.histplot(log_trans, kde=True)
        plt.title(f"Histogram after Log Transform: {col}")
        plt.show()
    else:
        print("Log transform skipped (column contains zero/negative values).")
```
